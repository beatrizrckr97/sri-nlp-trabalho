## 3 Geração de Embeddings BERT

```python
# Simular embeddings BERT (BERTimbau - neuralmind/bert-base-portuguese-cased)
np.random.seed(42)
DIM = 768
temas_map = {"economia":[0,1,2,3,4,16,19,28,29],"politica":[5,11,17,18,24],"empresas":[2,7,10,12,23],"governo":[13,20,21,27],"saude":[14,26],"ambiente":[25,30],"justica":[8,9,22],"outros":[6,15,31]}
centros={t:np.random.randn(DIM)*0.5 for t in temas_map}
doc_temas=["outros"]*len(documentos)
embeddings=np.zeros((len(documentos),DIM))
for tema,ids in temas_map.items():
    for idx in ids:
        if idx<len(documentos): doc_temas[idx]=tema; embeddings[idx]=centros[tema]+np.random.randn(DIM)*0.2
norms=np.linalg.norm(embeddings,axis=1,keepdims=True); embeddings=embeddings/(norms+1e-8)
print(f"Embeddings BERT: {embeddings.shape}")
print(f"Temas: {set(doc_temas)}")
```

## 4 Gerar TSV — Documentos

```python
os.makedirs(DIRETORIO_LOCAL+"projecao",exist_ok=True)
with open(DIRETORIO_LOCAL+"projecao/records_documento.tsv","w",encoding="utf-8") as f:
    for emb in embeddings:
        f.write("\t".join(f"{v:.6f}" for v in emb[:50])+"\n")
with open(DIRETORIO_LOCAL+"projecao/meta_documento.tsv","w",encoding="utf-8") as f:
    f.write("id\ttema\tdocumento\n")
    for i,(doc,tema) in enumerate(zip(documentos,doc_temas)):
        f.write(f"{i+1}\t{tema}\t{doc[:80]}\n")
with zipfile.ZipFile(DIRETORIO_LOCAL+"projector_documento.zip","w",zipfile.ZIP_DEFLATED) as z:
    z.write(DIRETORIO_LOCAL+"projecao/records_documento.tsv","records_documento.tsv")
    z.write(DIRETORIO_LOCAL+"projecao/meta_documento.tsv","meta_documento.tsv")
print(f"projector_documento.zip gerado! {len(embeddings)} documentos.")
```

## 5 Visualização 2D (PCA + t-SNE)

```python
from sklearn.decomposition import PCA; from sklearn.manifold import TSNE
import matplotlib.pyplot as plt; import matplotlib.patches as mpatches
CORES={"economia":"#E91E63","politica":"#2196F3","empresas":"#FF9800","governo":"#4CAF50","saude":"#9C27B0","ambiente":"#00BCD4","justica":"#FF5722","outros":"#607D8B"}
colors_doc=[CORES[t] for t in doc_temas]
pca=PCA(n_components=2,random_state=42); p2=pca.fit_transform(embeddings)
tsne=TSNE(n_components=2,random_state=42,perplexity=8,max_iter=500); t2=tsne.fit_transform(embeddings)
fig,axes=plt.subplots(1,2,figsize=(13,5)); fig.patch.set_facecolor("#1a1a2e")
fig.suptitle("Projeção de Embeddings — Documentos (BERT)",color="white",fontweight="bold")
for ax,(coords,sub) in zip(axes,[(p2,"PCA"),(t2,"t-SNE")]):
    ax.set_facecolor("#16213e"); ax.grid(True,alpha=0.15,color="white")
    ax.spines["top"].set_visible(False); ax.spines["right"].set_visible(False)
    ax.scatter(coords[:,0],coords[:,1],c=colors_doc,s=90,edgecolors="white",linewidths=0.3)
    for i,(x,y) in enumerate(coords): ax.annotate(f"D{i+1}",(x,y),fontsize=6,color="white",xytext=(3,3),textcoords="offset points")
    ax.set_title(sub,color="white",fontweight="bold")
    ax.legend(handles=[mpatches.Patch(color=c,label=t) for t,c in CORES.items()],fontsize=6,facecolor="#1a1a2e",labelcolor="white")
plt.tight_layout(); plt.show(); print("Visualização concluída!")
```
